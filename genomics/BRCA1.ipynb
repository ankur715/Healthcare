```python
gene = open("BRCA1.txt", "r")
```

```python
# characters
gene.read(1000)
```

```python
# lines
gene.readlines()
```

```python
a = 0
c = 0
g = 0
t = 0

for line in gene:
    line = line.lower()
    for char in line:
        if char == "a":
            a += 1
        if char == "c":
            c += 1
        if char == "g":
            g += 1
        if char == "t":
            t += 1
```

```python
print("number of a:", str(a))
print("number of c:", str(c))
print("number of g:", str(g))
print("number of t:", str(t))
```

```python
gc = (g+c+0.) / (a+c+g+t)
gc
```

```python
print("gc content:", gc)
```
